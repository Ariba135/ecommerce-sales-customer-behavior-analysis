```python
import pandas as pd

customers_df = pd.read_csv("data/processed/customers_cleaned.csv")

print("Customers dataset loaded successfully!")
print("Total records:", len(customers_df))
customers_df.head()
```

```python
customers_df.to_sql(
    name="customers",
    con=engine,
    if_exists="append",
    index=False
)

print("Customers dataset imported successfully!")
print("Total records imported:", len(customers_df))
```

```python
orders_df = pd.read_csv("data/processed/orders_cleaned.csv")

print("Orders dataset loaded successfully!")
print("Total records:", len(orders_df))

orders_df.head()
```

```python
orders_df.to_sql(
    name="orders",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=5000
)

print("Orders dataset imported successfully!")
print("Total records imported:", len(orders_df))
```
